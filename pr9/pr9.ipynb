```python
import pandas as pd
```

### Task 1


```python
from scipy.stats import chi2_contingency


def check_categorical_dependency(data, var1, var2, alpha=0.05):
    contingency_table = pd.crosstab(data[var1], data[var2])

    _, p_value, *_ = chi2_contingency(contingency_table)

    return p_value < alpha
```

```python
grants_data = pd.read_csv('data/grants.csv')

grants_data.head()
```

```python
print("Dependency between field and grant status:")
field_result = check_categorical_dependency(grants_data, 'field', 'status')
print("Features are dependent" if field_result else "Features are independent")
```

```python
grants_data['age_group'] = pd.cut(grants_data['oldest_age'],
                                  bins=[0, 30, 40, 50, 60, 100],
                                  labels=['<30', '30-40', '40-50', '50-60', '>60'])

print("Dependency between age group and grant status:")
age_result = check_categorical_dependency(grants_data, 'age_group', 'status')
print("Features are dependent" if age_result else "Features are independent")
```

### Task 2

```python
from scipy.stats import ttest_rel


def check_paired_samples_dependency(data, var1, var2, alpha=0.05):
    _, p_value = ttest_rel(data[var1], data[var2])

    return p_value < alpha
```

```python
medicine_data = pd.read_csv('data/med.csv')

medicine_data.head()
```

```python
print("Comparing blood pressure before and after treatment:")
result = check_paired_samples_dependency(medicine_data, 'Pressure_before', 'Pressure_after')
print("Features are dependent" if result else "Features are independent")
```
