```python
import pendulum
import pandas as pd
```

### Task 1

```python
dates = [
    "January 20, 2024",
    "2024-Apr-01",
    "11-Nov-2024",
    "08/12/10",
    "08/12/10",
    "08/12/10",
    "2024-11-11 11:11:01",
    "2024/12/24 07:30",
    "31/12/24 12:45:59",
    "8:05 01/01/2024",
    "9:20 29/02/2024",
    "12:30 24/12/2024",
    "23:58 30/12/2024"
]

for date_str in dates:
    try:
        parsed_date = pendulum.parse(date_str, strict=False)
        print(f"\nOriginal: {date_str}")
        print(f"Date (Date): {parsed_date}")
        print(f"POSIX Timestamp: {parsed_date.int_timestamp}")
    except Exception as e:
        print(f"Error parsing {date_str}: {e}")
```

### Task 2

```python
df = pd.read_excel('df_10.xlsx')

df.head()

df['dob'] = pd.to_datetime(df[['year', 'month', 'day']])
df['tob'] = pd.to_datetime(
    dict(
        year=df['year'],
        month=df['month'],
        day=df['day'],
        hour=df['hour'],
        minute=df['min']
    )
)

df['dob_wd'] = df['dob'].dt.day_name()

current_date = pendulum.now()
df['age_fy'] = df['dob'].apply(lambda x: current_date.diff(pendulum.instance(x)).in_years())
```
